```python
# Indexing - Python Starts at Zero

rainfall=[10,25,32,18,45]
print(rainfall[0])
print(rainfall[1])
```

```python
# Negative Indexing

rainfall=[10,25,32,18,45]
print(rainfall[-1])
print(rainfall[-2])
```

```python
# Modifying a List - Mutability

rainfall=[10,25,32,18,45]
rainfall[1] = 30
print(rainfall)
```

```python
# Adding (append, insert) & Removing Items (remove, pop)

rainfall=[10,25,32,18,45]
rainfall.append(50)
rainfall.insert(2,40)
rainfall.remove(40)
rainfall.pop()
print(rainfall)
```
