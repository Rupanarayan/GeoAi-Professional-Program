```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt

landsat = rasterio.open("landsat.tif")

print("Width:", landsat.width)
print("Height:", landsat.height)
print("Number of bands:", landsat.count)
```

```python
print("CRS:", landsat.crs)
print("Resolution:", landsat.res)
print("Pixel area:", 
      landsat.res[0] * landsat.res[1], "m²")
print(landsat.bounds)
print(landsat.transform)
```

```python
print(landsat.meta)
print(landsat.dtypes)
print(landsat.nodata)
```

```python
band1 = landsat.read (1, masked=True)
print(type(band1))
print(band1.shape)
```

```python
plt.figure(figsize=(8, 6))
plt.imshow(band1)
plt.colorbar(label="Pixel Value")
plt.title("Chandigarh Landsat - Band 1")
plt.show()
```

```python
modis = rasterio.open("modis.tif")
sentinel5p = rasterio.open("sentinel_5p.tif")
print("Landsat:", landsat.width,
      landsat.height, landsat.count, landsat.res)
print("Modis:", modis.width,
      modis.height, modis.count, modis.res)
print("Sentinel-5p:", sentinel5p.width,
      sentinel5p.height, sentinel5p.count, sentinel5p.res)
```
