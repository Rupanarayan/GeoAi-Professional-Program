```python
import pandas as pd
import matplotlib.pyplot as pit

df=pd.read_csv("Day7.csv")
```

```python
plt.plot(
    delhi["date"],
    delhi["temperature"],
    marker="^"
)
plt.show
```

```python
delhi=df[df["station_name"]=="Delhi"]
```

```python
plt.figure(figsize=(8,4))

plt.plot(
    delhi["date"],
    delhi["temperature"],
    color="red",
    linestyle="--",
    marker="o",
    linewidth=2,
    markersize=6
)

plt.xlabel("Date")
plt.ylabel("Temperature (C)")
plt.title("Daily Temerature - Delhi")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.hist(
    df["temperature"],
    bins=8,
    color="skyblue",
    edgecolor="black"
)

plt.xlabel("Temperature(c)")
plt.ylabel("Frequency")
plt.title("Temperature Distribution")

plt.show
```

```python
df.boxplot(
    column="temperature",
    by="station_name"
)

plt.xlabel("Station")
plt.ylabel("Temperature")
plt.title("Temperature Distribution by station")

plt.show
```

```python
plt.plot(df["date"], df["temperature"])
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.title("Daily Temperature")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.plot(df["date"], df["temperature"],
color="red", linestyle="--")
plt.show()
```

```python
temp_region = df.groupby("region")["temperature"].mean()
temp_region
temp_region.plot(kind="bar", color="orange")
plt.xlabel("Region")
plt.ylabel("Average Temperature (°C)")
plt.title("Average Temperature by Region")
plt.show()
```

```python
plt.scatter(df["temperature"], df["ndvi"],
color="green", marker="o", s=60)
plt.xlabel("Temperature (°C)")
plt.ylabel("NDVI")
plt.title("Temperature vs NDVI")
plt.show()
```

```python
plt.hist(df["temperature"], bins=10,
color="skyblue", edgecolor="black")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.title("Temperature Distribution")
plt.show()
```

```python
df.boxplot(column="temperature", by="region")
plt.xlabel("Region")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Distribution by Region")
plt.suptitle("")
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["date"], df["temperature"],
label="Temperature", marker="o")
plt.plot(df["date"], df["humidity"],
label="Humidity", marker="s")
plt.legend()
plt.show()
```
